```python
a=pd.read_csv("products-100 (1).csv")
a
```

```python
le= LabelEncoder()
a['Availability']=le.fit_transform(
    a['Availability']
)
```

```python
x=a[['Internal ID','Price']]
y=a['Price']
```

```python
x_train,x_test,y_train,y_test=train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
r2=r2_score(y_test,y_pred)
print("\n" + "=" * 40)
print("MODEL VALIDATION")
print("=" * 40)

print("R2 Score :", round(r2, 4))
```

```python
print("\nIntercept")
print(round(model.intercept_, 2))
print("\nCoefficients")

for feature, coef in zip(x.columns, model.coef_):
  print(feature,":",round(coef,2))
```

```python
import pandas as pd

# Get user inputs safely using valid variable names (no spaces)
internal_id = float(input("Enter internal ID: "))
price_val = float(input("Enter price: "))

# Create a DataFrame using the exact feature names the model was trained on
user_data = pd.DataFrame(
    [[internal_id, price_val]],
    columns=["Internal ID", "Price"]
)

# Make prediction
prediction = model.predict(user_data)

print("\n" + "=" * 40)
print("PREDICTION RESULT")
print("=" * 40)
print("Internal ID:", internal_id)
print("Price Input:", price_val)
print("Predicted Output:", round(prediction[0], 2))
```
