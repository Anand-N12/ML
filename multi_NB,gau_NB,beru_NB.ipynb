```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.preprocessing import LabelEncoder
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
emails=[
    "win money",
    "free prize",
    "hello friend",
    "how are you"
]

labels=["spam","spam","Not spam","Not spam"]

le=LabelEncoder()
y=le.fit_transform(labels)

vectorizer=CountVectorizer()
x=vectorizer.fit_transform(emails)

model=MultinomialNB()
model.fit(x,y)
test_email=["win prize"]
test_email_vectorized=vectorizer.transform(test_email)
prediction=model.predict(test_email_vectorized)
print(prediction)

sns.heatmap(pd.DataFrame(x.toarray()),annot=True)
```

```python
import numpy as np
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder
import pandas as pd
```

```python
X = np.array([[40], [45], [50], [60], [70]])
y_text = ["Fail", "Fail", "Fail", "Pass", "Pass"]
```

```python
le = LabelEncoder()
y = le.fit_transform(y_text)
```

```python
model = GaussianNB()
```

```python
model.fit(X, y)

new_marks = [[55]]
prediction = model.predict(new_marks)
print("Predicted for 55 marks:", le.inverse_transform(prediction)[0])
```

```python
from sklearn.naive_bayes import BernoulliNB
from sklearn.preprocessing import LabelEncoder
```

```python
X = np.array([
    [1, 1],
    [1, 0],
    [0, 1],
    [0, 0]
])

y_text = ["Yes", "Yes", "No", "No"]
```

```python
le = LabelEncoder()
y = le.fit_transform(y_text)
```

```python
bernoulli_model = BernoulliNB()
```

```python
bernoulli_model.fit(X, y)
new_dish=[[1,0]]
prediction=bernoulli_model.predict(new_dish)
print("Predicted for new dish:",le.inverse_transform(prediction)[0])
```
